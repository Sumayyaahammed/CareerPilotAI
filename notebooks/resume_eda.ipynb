```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load Resume Dataset

resume_df = pd.read_csv("../data/resumes/Resume.csv")

# Display first 5 rows
resume_df.head()

```

```python
print("Dataset Shape:")
print(resume_df.shape)

print("\nColumn Names:")
print(resume_df.columns)

print("\nDataset Information:")
resume_df.info()
```

```python
#check missing values
print("Missing Values:\n")

print(resume_df.isnull().sum())
```

```python
#check duplicates

duplicates = resume_df.duplicated().sum()

print("Duplicate Records:", duplicates)
```

```python
resume_df = resume_df.drop_duplicates()

print("Dataset Shape After Removing Duplicates:")
print(resume_df.shape)
```

```python
print(resume_df.dtypes)
```

```python
category_counts = resume_df["Category"].value_counts()

print(category_counts)
```

```python
plt.figure(figsize=(12,6))

category_counts.plot(kind="bar")

plt.title("Resume Categories")

plt.xlabel("Category")

plt.ylabel("Count")

plt.xticks(rotation=90)

plt.tight_layout()

plt.show()
```

```python
resume_df["Resume_Length"] = resume_df["Resume_str"].apply(len)

resume_df.head()
```

```python
resume_df["Resume_Length"].describe()
```

```python
plt.figure(figsize=(10,5))

plt.hist(resume_df["Resume_Length"], bins=30)

plt.title("Distribution of Resume Lengths")

plt.xlabel("Resume Length")

plt.ylabel("Frequency")

plt.show()
```

```python
longest = resume_df["Resume_Length"].max()

print("Longest Resume Length:", longest)
```

```python
shortest = resume_df["Resume_Length"].min()

print("Shortest Resume Length:", shortest)
```

```python
average = resume_df["Resume_Length"].mean()

print("Average Resume Length:", average)
```

```python
resume_df.to_csv(
    "../data/processed/resume_cleaned.csv",
    index=False
)
```
